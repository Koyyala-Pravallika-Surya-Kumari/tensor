```python
        #concetanation
import torch
a=torch.tensor([[1,2,3],[4,5,6]])
b=torch.tensor([[7,8,9],[1,3,4]])
c=torch.cat((a,b),dim=0)
d=torch.cat((a,b),dim=1)
print(c)
print(d)
```

```python
#operations
import torch
a=torch.tensor([[5,6,7],[8,9,10]],dtype=float)
print(a.sum())
print(a.sum(dim=0))
print(a.sum(dim=1))
print(a.mean(dim=0))
print(a.mean(dim=1))
```

```python
import torch
a=torch.tensor([[5,6,7],[8,9,10]])
print(a.max())
print(a.max(dim=0))
print(a.max(dim=1))
```
